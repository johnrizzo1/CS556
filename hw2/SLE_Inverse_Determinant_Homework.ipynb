## 1. Solving System of Linear Equations

### 1.1 Row Reduction approach

Solve the following system of linear equations using the reduction method (as mentioned in the week 2 lab):
$$\begin{cases}
x + 2y - 3z + 4w = 12, \\ 2x + 2y - 2z + 3w = 10, \\ y + z = -1, \\ x - y + z - 2w = -4 \end{cases}\tag{1}$$


```python
import numpy as np
```

### 1.2 `Numpy` linear algebra package
Solve the above system of linear equations using the `numpy.linalg` package.

```python
A = np.array([
    [1, 2, -3, 4],
    [2, 2, -2, 3],
    [0, 1, 1, 0],
    [1, -1, 1, -2]
])

b = np.array([12, 10, -1, -4])

print(np.linalg.solve(A, b))
```

# 3. Matrix Inverse
### 3.1 Compute the inverse of non-singular matrix $B$ using row reduction
$$ \begin{align}
  \textbf{B}  = \begin{bmatrix}
1 & 2 & 1\\
4 & 4 & 5\\
6 & 7 & 7
\end{bmatrix}
  \end{align}
  $$

```python
def MultiplyRow(M, row, scalar):
    M_new = M.copy()
    M_new[row] = M_new[row] * scalar
    return M_new

def AddRows(M, row_num_1, row_num_2, scalar):
    M_new = M.copy()
    M_new[row_num_2] = scalar * M_new[row_num_1] + M_new[row_num_2]
    return M_new

def ExchangeRows(M, row_num_1, row_num_2):
    M_new = M.copy()
    M_new[[row_num_1, row_num_2]] = M_new[[row_num_2, row_num_1]]
    return M_new
```

```python
B = np.array([
    [1, 2, 1],
    [4, 4, 5],
    [6, 7, 7]
], dtype=np.dtype(float))

B_system = np.hstack((B, np.eye(3,3))); print(B_system)
B_system = AddRows(B_system, 0, 1, -4); print(B_system)
B_system = AddRows(B_system, 0, 2, -6); print(B_system)
B_system = MultiplyRow(B_system, 1, -1/4); print(B_system)
B_system = AddRows(B_system, 1, 2, 5); print(B_system)
B_system = MultiplyRow(B_system, 2, -4); print(B_system)
B_system = AddRows(B_system, 2, 1, .25); print(B_system)
B_system = AddRows(B_system, 2, 0, -1); print(B_system)
B_system = AddRows(B_system, 1, 0, -2); print(B_system)
```

```python
# extract inverse from computed matrix
B_inv = B_system[:, -3:]; print(B_inv)
```

```python
# Test if B*B_inv = I
B.dot(B_inv)
```
